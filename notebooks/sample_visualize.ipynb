# Visualization Traffic Sign Samples

```python
# Import essential libraries
import os
import sys
import torch
import numpy as np
import matplotlib
# 'Agg' prevents Matplotlib from looking for a monitor (prevents freezes) when running on remote servers
matplotlib.use('Agg') 

import matplotlib.pyplot as plt
# specific command to Jupyter notebook to display plots inline
%matplotlib inline

# start from /notebooks dir to /src dir
root_path = os.path.abspath("..")
if root_path not in sys.path:
    sys.path.append(root_path)

from src.Data.Data2 import ImageDataset2

print("Libraries imported successfully.")
```

## 1. Loading Dataset

```python
data_dir = os.path.join(root_path, "data")
image_dir = os.path.join(data_dir, "GTSRB", "Final_Training", "Images")
csv_file = os.path.join(data_dir, "concepts_per_class.csv")

img_size = 128

dataset = ImageDataset2(image_dir, csv_file, img_size, img_size)
print(f" Dataset loaded successfully.")
print(f" Image size: {img_size} x {img_size}")
print(f" Number of images: {len(dataset)}")
```

## 2. Visualize Random Samples

```python
# Visualize some random traffic sign samples
def visualize_samples(dataset, num_samples=8, title="Traffic Sign Samples"):
    fig, axes = plt.subplots(2, 4, figsize=(16, 8))
    axes = axes.ravel() 

    sample_list = torch.randperm(len(dataset))[:num_samples]

    for i, idx in enumerate(sample_list):
        
        img, (vector, label) = dataset[idx]
            
        # transform Tensor dimension (C, H, W) with numpy to (H, W, C) for Matplotlib
        img_np = img.permute(1, 2, 0).numpy()

        axes[i].imshow(img_np)           
        
        label_id = label.item()
        label_name = dataset.__getname__(label_id)
      
        axes[i].set_title(f"Class_id: {label_id}\nClass_name: {label_name}", fontsize=9)
        axes[i].axis('off')

    plt.suptitle(title, fontsize=16)
    plt.tight_layout()
    plt.show()

visualize_samples(dataset)
```
